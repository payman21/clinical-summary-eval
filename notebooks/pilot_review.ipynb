# Pilot review

Read the 30 pilot notes section by section to settle the Stage 0 scope question:
**Brief Hospital Course only** vs. **BHC + discharge sections**.

Kernel: project `.venv`. Outputs are stripped on commit (nbstripout), so note text never lands in git.

```python
import pandas as pd

from clinsum.data import load_pilot, section_lengths, split_sections

pd.set_option("display.max_columns", 30, "display.width", 200)

pilot = load_pilot()
pilot[["note_id", "n_chars", "length_q", "service", "service_group"]]
```

## Section lengths (characters)

```python
lengths = section_lengths(pilot)

discharge_sections = ["Discharge Medications", "Discharge Diagnosis", "Discharge Instructions"]
scope = pd.DataFrame({
    "full_note": lengths.sum(axis=1),
    "bhc": lengths["Brief Hospital Course"],
    "bhc_plus_discharge": lengths["Brief Hospital Course"] + lengths[discharge_sections].sum(axis=1),
})
scope.describe().round(0)
```

```python
lengths.describe().T[["mean", "50%", "min", "max"]].round(0)
```

## Read one note, section by section

Change `i` (0–29) and re-run. Narrow `SHOW` to focus on the sections relevant to the scope decision.

```python
SHOW = None  # e.g. ["Brief Hospital Course", "Discharge Medications", "Discharge Diagnosis", "Discharge Instructions"]


def show_note(i: int, show=SHOW):
    row = pilot.iloc[i]
    print(f"[{i}] {row.note_id}  |  {row.service} ({row.service_group})  |  {row.length_q}  |  {row.n_chars} chars")
    for name, body in split_sections(row.text).items():
        if show is None or name in show:
            print(f"\n{'=' * 10} {name.upper()} ({len(body)} chars) {'=' * 10}\n{body}")


i = 3
show_note(i)
```
